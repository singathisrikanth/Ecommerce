```python
print("pipeline started")

raw_records = 10000
duplicate_records = 200
null_records= 150
print("data extracted")
print(f"raw records {raw_records}")

print("Data cleaned")

clean_records = raw_records - duplicate_records -null_records
print(f"cleaned_records {clean_records}")

# Step 3: Calculate quality score

quality_score = (clean_records/raw_records)* 100
print(f"quality_data {quality_score}")


# Step 4: Decision

if quality_score >= 90:
 print("Data is good - loading to warehosue")
else:
   print("Data need improvements to load into warehosue")
```

```python
# Pipeline Health Checker
print("="*40)
print(" Pipeline Health Checker")
print("="* 40)

pipeline_name ="sales Data Pipeline"
source = "Mysql Database"
desintation = "Databricks warehouse"

total_records = 50000
failed_records = 300
processing_time = 45.7
max_allowed_time = 60

#calculations

success_records = total_records-failed_records
success_rate  = (success_records/total_records)*100
failure_rate = (failed_records/total_records)*100


print(f"Pipeline :{pipeline_name}")
print(f" source : {source}")
print(f" desintation : {desintation}")
print(f" total_records : {total_records}")
print(f" success_records : {success_records}")
print(f" success_rate : {success_rate}")
print(f" failure_rate : {failure_rate}")


if success_rate >= 90:
   print("Data Quality :  Excellent")
elif success_rate >= 85:
   print("Data Quality: AVERAGE - needs attention")
else:
  print("Data Quality: POOR - pipeline stopped!")


if processing_time <= max_allowed_time:
  print("Processing Time: ON TIME")
else:
   print("Processing Time: TOO SLOW - needs optimization!")

if success_rate >= 95 or success_rate <= max_allowed_time:
  print("PIPELINE STATUS: HEALTHY!")
else:
   print("PIPELINE STATUS: NEEDS ATTENTION!")

print("="*40)
```
